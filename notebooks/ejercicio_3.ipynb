# Ejercicio 3

```python
import sys
sys.path.append("../src")

import numpy as np
import pandas as pd

from lista4.data import load_csv
from lista4.models import fit_linear
from lista4.metrics import riesgo_cuadratico
```

```python
df = load_csv("experimento_3.csv")
x1 = df["x1"].values
x2 = df["x2"].values
y = df["y"].values
z2 = x2 / 100.0

X_original = np.column_stack([x1, x2])
X_reescalado = np.column_stack([x1, z2])
```

## (a) Coeficientes y riesgos de los dos modelos

```python
modelo_original = fit_linear(X_original, y)
modelo_reescalado = fit_linear(X_reescalado, y)

R_original = riesgo_cuadratico(y, modelo_original.predict(X_original))
R_reescalado = riesgo_cuadratico(y, modelo_reescalado.predict(X_reescalado))

print("Modelo original:    beta0 =", modelo_original.intercept_,
      " beta1 =", modelo_original.coef_[0], " beta2 =", modelo_original.coef_[1],
      " R =", R_original)
print("Modelo reescalado:  gamma0 =", modelo_reescalado.intercept_,
      " gamma1 =", modelo_reescalado.coef_[0], " gamma2 =", modelo_reescalado.coef_[1],
      " R =", R_reescalado)
```

## (b) Diferencia máxima entre predicciones

```python
y_pred_1 = modelo_original.predict(X_original)
y_pred_2 = modelo_reescalado.predict(X_reescalado)

d = np.abs(y_pred_1 - y_pred_2)
max_d = np.max(d)
print("max_i d_i =", max_d)
```

## (c) Verificación numérica de la relación entre coeficientes

```python
error_beta0 = abs(modelo_reescalado.intercept_ - modelo_original.intercept_)
error_beta1 = abs(modelo_reescalado.coef_[0] - modelo_original.coef_[0])
error_beta2 = abs(modelo_reescalado.coef_[1] - 100 * modelo_original.coef_[1])

print("|gamma0 - beta0|       =", error_beta0)
print("|gamma1 - beta1|       =", error_beta1)
print("|gamma2 - 100*beta2|   =", error_beta2)
```

## (d) Comparación de |beta2| y |gamma2|, y de las predicciones

```python
print("|beta2|  =", abs(modelo_original.coef_[1]))
print("|gamma2| =", abs(modelo_reescalado.coef_[1]))
```

Comparación: $|\hat\beta_2|\approx0.050$ y $|\hat\gamma_2|\approx5.048$ son muy distintos, una diferencia de exactamente 100 veces, como se espera al reescalar $x_2\to z_2=x_2/100$. Sin embargo, al comparar las predicciones de ambos modelos, $\max_i d_i\approx8.9\times10^{-15}$, las predicciones de ambos modelos son prácticamente idénticas. Esto confirma que el reescalamiento de una variable no cambia lo que el modelo predice, solo cambia la magnitud del coeficiente asociado para compensar la nueva escala.

```python
tabla_3 = pd.DataFrame({
    "modelo": ["original (x2)", "reescalado (z2=x2/100)"],
    "intercepto": [modelo_original.intercept_, modelo_reescalado.intercept_],
    "coef_x1": [modelo_original.coef_[0], modelo_reescalado.coef_[0]],
    "coef_x2_o_z2": [modelo_original.coef_[1], modelo_reescalado.coef_[1]],
    "riesgo_empirico": [R_original, R_reescalado],
    "max_i d_i": [max_d, max_d],
})
tabla_3
```
