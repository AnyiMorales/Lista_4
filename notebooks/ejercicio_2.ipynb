```python
import sys
sys.path.append("../src")

import numpy as np
import pandas as pd

from lista4.data import load_csv
from lista4.models import fit_linear
from lista4.metrics import riesgo_cuadratico
```

## (a)-(c) Ajuste, correlaciones y tabla resumen

```python
archivos = ["experimento_2_a.csv", "experimento_2_b.csv", "experimento_2_c.csv"]

filas = []
for nombre in archivos:
    df = load_csv(nombre)
    x1 = df["x1"].values
    x2 = df["x2"].values
    y = df["y"].values
    X12 = df[["x1", "x2"]].values

    corr_x1_x2 = np.corrcoef(x1, x2)[0, 1]

    modelo_x1 = fit_linear(x1, y)
    modelo_x12 = fit_linear(X12, y)

    R_x1 = riesgo_cuadratico(y, modelo_x1.predict(x1.reshape(-1, 1)))
    R_x12 = riesgo_cuadratico(y, modelo_x12.predict(X12))

    filas.append({
        "archivo": nombre,
        "corr(x1,x2)": corr_x1_x2,
        "R_x1": R_x1,
        "R_x1_x2": R_x12,
        "beta1": modelo_x12.coef_[0],
        "beta2": modelo_x12.coef_[1],
    })

tabla_2 = pd.DataFrame(filas)
tabla_2
```

## (d) Delta R por archivo

```python
tabla_2["delta_R"] = tabla_2["R_x1"] - tabla_2["R_x1_x2"]
tabla_2
```

## (e) Rango de beta1, beta2 y delta_R entre los tres archivos

```python
def rango(z):
    return np.max(z) - np.min(z)

rango_beta1 = rango(tabla_2["beta1"].values)
rango_beta2 = rango(tabla_2["beta2"].values)
rango_deltaR = rango(tabla_2["delta_R"].values)

print("rango(beta1) =", rango_beta1)
print("rango(beta2) =", rango_beta2)
print("rango(delta_R) =", rango_deltaR)
```

## (f) Comparación

```python
print(tabla_2[["archivo", "corr(x1,x2)", "R_x1", "R_x1_x2", "delta_R", "beta1", "beta2"]])
print()
print("rango(delta_R) =", rango_deltaR)
print("rango(beta1)   =", rango_beta1)
print("rango(beta2)   =", rango_beta2)
```
